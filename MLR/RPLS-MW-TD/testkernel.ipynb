```python
import pandas as pd

# Đọc file CSV với dấu phân cách ";"
df = pd.read_csv(
    "AirQualityUCI.csv",  # <-- đổi tên file nếu cần
    sep=";",
    decimal=","           # số thập phân dùng dấu phẩy (kiểu châu Âu)
)

# Xoá cột rỗng ở cuối (nếu có)
df = df.dropna(axis=1, how="all")

# Gộp Date + Time thành cột DateTime, rồi parse
df["DateTime"] = pd.to_datetime(
    df["Date"] + " " + df["Time"],
    format="%d/%m/%Y %H.%M.%S"
)

# Chuyển sang Unix timestamp (số nguyên, đơn vị: giây)
df["DateTime"] = df["DateTime"].astype("int64") // 10**9

# Xoá cột Date và Time gốc
df.drop(columns=["Date", "Time"], inplace=True)

# Đưa cột DateTime lên đầu
cols = ["DateTime"] + [c for c in df.columns if c != "DateTime"]
df = df[cols]

# Lưu kết quả
output_file = "AirQualityUCI_fixed.csv"
df.to_csv(output_file, sep=";", index=False)

print(f"✅ Đã lưu: {output_file}")
print(df.head())
```
